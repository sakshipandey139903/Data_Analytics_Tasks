```python
import pandas as pd

data = {
    'Name': ['  RAHUL  ', 'priya ', ' AMIT', 'Sneha  ', 'rahul'],
    'City': ['Delhi ', ' MUMBAI', 'delhi', 'Bangalore ', 'Delhi'],
    'Email': ['RAHUL@GMAIL.COM', 'priya@gmail.com', 'amit@yahoo.com', None, 'rahul@gmail.com']
}

df = pd.DataFrame(data)
print(df)
```

```python
df['Name'] = df['Name'].str.strip().str.lower()
df['City'] = df['City'].str.strip().str.lower()
df['Email'] = df['Email'].str.lower()
print(df)
```

```python
df['City'] = df['City'].str.replace('delhi', 'new delhi')
print(df)

```

```python
import pandas as pd
import numpy as np

data = {
    'Name': ['  RAHUL  ', 'priya ', ' AMIT', 'Sneha  ', 'rahul'],
    'City': ['Delhi ', ' MUMBAI', 'delhi', 'Bangalore ', 'Delhi'],
    'Email': ['RAHUL@GMAIL.COM', 'priya@gmail.com', 'amit@yahoo.com', None, 'rahul@gmail.com'],
    'Age': [20, None, 22, 200, 19],
    'Marks': [85, 90, None, 78, 95]
}

df = pd.DataFrame(data)
print(df)
```

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
print(df)
```

```python
df['Marks'] = df['Marks'].fillna(df['Marks'].median())
print(df)
```

```python
df['City'] = df['City'].fillna(df['City'].mode()[0])
print(df)
```

```python
df['Age'] = df['Age'].round(0).astype(int)
df['Marks'] = df['Marks'].astype(int)
print(df)
```
